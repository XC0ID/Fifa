## Imoprting Basic Lib & Load Data

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings("ignore")
```

```python
df=pd.read_csv("male_players_19.csv")
pd.set_option("display.max_columns",None)
df.head()
```

### DOMAIN ANALYSIS

- The dataset contains information about total number of football players is  18,278.
- The average age of players is around 25 years, with a minimum age of 16 and a maximum age of 42.
- The average height is approximately 181 cm, with a range from 156 cm to 205 cm.
- The average weight is about 75 kg, ranging from 50 kg to 110 kg.
- The players' overall ratings have an average of about 66, with a potential rating averaging around 71.
- The highest-rated player has an overall rating of 94, indicating an exceptional talent.
- The lowest overall rating is 48, which suggests lower-tier players.
- The average player value is 2.5 million euros, with a wide range of values, including some players valued as high as 105.5 million euros.
- Player wages vary, with an average wage of around 9,457 euros per week.
- Skill moves and weak foot ratings are mostly concentrated around 3, indicating that many players have moderate skills in this regard.
- Release clauses have a wide range, with an average of around 4.7 million euros.
- Most player contracts are valid until the year 2021, but there is variability, with some extending beyond that year.

## Graphical Visualiztions

```python
numeric_cols = ['overall', 'potential', 'age', 'height_cm', 'weight_kg', 'pace', 'shooting', 'passing', 'dribbling', 'defending', 'physic']
corr = df[numeric_cols].corr()
plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, cmap='Blues', fmt='.2f')
plt.title('Correlation Heatmap of Key Player Attributes')
plt.show()
```

```python
key_attributes = ['pace', 'shooting', 'passing', 'dribbling', 'defending', 'physic']
sns.pairplot(df[key_attributes], diag_kind='kde',  plot_kws={'color': 'lightcoral'},diag_kws={'color': 'lightcoral'} )
plt.suptitle('Pair Plot of Key Player Attributes', y=1.02)
plt.show()
```

```python
columns_to_plot = ["age", "weight_kg", "height_cm","power_strength","power_stamina","power_jumping"]
plt.figure(figsize=(15,12))
plotnumber=1

for column in columns_to_plot:
    if plotnumber<8:
        plt.subplot(3,3,plotnumber)
        sns.lineplot(x=df[column], y=df["physic"], label=column,color="orange",linewidth = 2)
        sns.set_style("darkgrid", {"grid.color": ".7", "grid.linestyle": ":"})
        plt.xlabel(column, fontsize=15)
        plt.ylabel("Physic", fontsize=15)
    plotnumber+=1
plt.tight_layout()
plt.legend()
plt.show()
```

```python
columns_to_plot2 = ["skill_moves", "power_long_shots", "power_shot_power","weak_foot"]
plt.figure(figsize=(15,10))
plotnumber=1

for column in columns_to_plot2:
    if plotnumber<7:
        plt.subplot(2,2,plotnumber)
        sns.lineplot(x=df[column], y=df["shooting"], label=column, color="green",linewidth = 2)
        plt.xlabel(column, fontsize=15)
        plt.ylabel("shooting", fontsize=15)
    plotnumber+=1
plt.tight_layout()
plt.legend()
plt.show()
```

```python
columns_to_plot3 = ["potential","weak_foot","skill_moves","passing","dribbling",
                   "pace","power_stamina","power_strength"]
plt.figure(figsize=(20,15))
plotnumber=1

for column in columns_to_plot3:
    if plotnumber<12:
        plt.subplot(3,3,plotnumber)
        sns.lineplot(x=df[column], y=df["overall"], label=column,color="blue")
        plt.xlabel(column, fontsize=20)
        plt.ylabel("overall", fontsize=20)
    plotnumber+=1
plt.tight_layout()
plt.legend()
plt.show()
```

```python
columns_to_plot4 = ["skill_dribbling","skill_curve","skill_fk_accuracy","skill_long_passing","skill_ball_control"]
plt.figure(figsize=(16,16))
plotnumber=1

for column in columns_to_plot4:
    if plotnumber<20:
        plt.subplot(5,3,plotnumber)
        sns.lineplot(x=df[column], y=df["overall"], label=column,color="red")
        plt.xlabel(column, fontsize=15)
        plt.ylabel("overall", fontsize=15)
    plotnumber+=1
plt.tight_layout()
plt.legend()
plt.show()
```

```python
skills = ['pace','shooting','passing','dribbling','defending','physic']
means = df[skills].mean()  

plt.figure(figsize=(10,5))
colors = ['lightcoral', 'lightgreen', 'lightblue', 'lightsalmon', 'plum', 'tan'] 
plt.bar(skills, means, color=colors, edgecolor='black')
plt.title("Average Player Attributes")
plt.xlabel("Player Attributes")
plt.ylabel("Average Rating")
plt.xticks(rotation=45)
plt.show()
```

```python
corr = df.corr(numeric_only=True)['overall'].sort_values(ascending=False)[1:15]
plt.figure(figsize=(8,6))
colors = ['lightcoral', 'lightgreen', 'lightblue', 'lightsalmon', 'plum', 'tan', 'lightseagreen','lightyellow', 'lightpink', 'lightgray', 'lightcyan', 'lightsteelblue', 'lightgoldenrodyellow'] 
sns.barplot(x=corr.values, y=corr.index, palette=colors, edgecolor='black')
plt.title("Top Attributes Affecting Overall Rating")
plt.xlabel("Correlation with Overall Rating")
plt.ylabel("Attributes")
plt.show()
```
